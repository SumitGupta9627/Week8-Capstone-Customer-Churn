```python
import pandas as pd
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,roc_auc_score
df=pd.read_csv("../data/cleaned_data.csv")
print("Contract chi-square p:",stats.chi2_contingency(pd.crosstab(df.Contract,df.Churn))[1])
print("Tenure t-test p:",stats.ttest_ind(df.loc[df.Churn==1,"Tenure"],df.loc[df.Churn==0,"Tenure"],equal_var=False).pvalue)
X=df.drop(columns=["CustomerID","Churn","ChurnLabel","TenureBand"]); y=df.Churn
pre=ColumnTransformer([("cat",OneHotEncoder(handle_unknown="ignore"),["Contract","PaymentMethod","PaperlessBilling"]),("num",StandardScaler(),["Tenure","MonthlyCharges","TotalCharges","SeniorCitizen"])])
m=Pipeline([("pre",pre),("model",LogisticRegression(max_iter=1000,random_state=42))])
a,b,c,d=train_test_split(X,y,test_size=.2,random_state=42,stratify=y); m.fit(a,c); p=m.predict(b); q=m.predict_proba(b)[:,1]
print(accuracy_score(d,p),precision_score(d,p),recall_score(d,p),f1_score(d,p),roc_auc_score(d,q))
```
