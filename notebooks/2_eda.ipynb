# 2. Exploratory Data Analysis

Week 8 Capstone — Customer Churn Business Analysis

```python
import pandas as pd, matplotlib.pyplot as plt
df=pd.read_csv("../data/cleaned_data.csv")
print("Overall churn:",df.Churn.mean())
print(df.groupby("Contract").Churn.mean().sort_values(ascending=False))
print(df.groupby("TenureBand",observed=False).Churn.mean())
plt.figure(figsize=(8,4)); plt.bar(df.groupby("Contract").Churn.mean().index,df.groupby("Contract").Churn.mean().values); plt.title("Churn by Contract"); plt.ylabel("Churn Rate"); plt.show()
```
